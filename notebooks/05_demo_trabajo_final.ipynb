## 3. Consulta directa al RAG de MITRE ATT&CK (FAISS)

Una vez identificados los puertos y servicios expuestos, consulta en el RAG la técnica MITRE relacionada con el descubrimiento de servicios de red (Network Service Discovery)."

#### 1. Análisis de la ejecución (Lo que hizo el agente)

* Enrutamiento: El nodo router interceptó la solicitud y la derivó directamente al agente especialista infraestructura_redes.

* Uso de Tools: El agente ejecutó la herramienta infra_tools invocando nmap de forma segura (aplicando la allowlist de parámetros) sobre la IP 192.168.0.26 para auditar los puertos activos.

* Consulta RAG: Con los puertos identificados, realizó la búsqueda vectorial en el índice FAISS consultando por el concepto "Network Service Discovery".

* Síntesis: Correlacionó los hallazgos técnicos del escaneo activo con la ficha oficial de MITRE ATT&CK y entregó la descripción contextualizada del riesgo.

#### 2. Resultado Esperado

Se esperaba que el agente aislará el flujo en la vertical de red, ejecutara un escaneo válido sobre el host objetivo, identificara la técnica exacta de MITRE asociada al escaneo de servicios (T1046) y presentara una propuesta de mitigación orientada a infraestructura.

#### 3. Por qué funcionó correctamente

* Validación de Parámetros: La lista blanca de argumentos de nmap evitó que el LLM introdujera flags no permitidos o peligrosos durante el escaneo.

* Recuperación Semántica Fiel: El modelo de embeddings asoció la descripción de los servicios abiertos con la entrada vectorial de T1046 (Network Service Discovery), superando la limitación de coincidencias textuales rígidas.

#### 1. Análisis de la ejecución (Lo que hizo el agente)

* Enrutamiento Dinámico: El nodo router identificó la intención de análisis de eventos del SIEM y canalizó el flujo hacia el agente especializado soc_logs.

* Invocación Interactiva de Tools y RAG: Tal como refleja la traza de la consola, el agente ejecutó una secuencia coordinada de acciones:

  * Consultando Wazuh...: Invocó soc_tools para realizar consultas activas al Indexer/API de Wazuh sobre el agente ubuntuLocal (ID: 001).

  * Buscando en MITRE ATT&CK...: Ejecutó una búsqueda semántica en la base vectorial FAISS para mapear el comportamiento detectado.

  * Consultando Wazuh...: Reconsultó la telemetría histórica de auditd para extraer los eventos exactos de comandos ejecutados (/usr/bin/docker, /usr/bin/git, /usr/bin/uname, etc.).

  * Mapeo Táctico Identificado:

    * Acceso a Credenciales / Fuerza Bruta: Mapeado a T1110 (Brute Force).

    * Escalada de Privilegios: Mapeado a T1548.003 (Sudo and Sudo Caching) y T1068 (Exploitation for Privilege Escalation), asociándolo a la modificación anómala de /etc/sudoers.

* Exportación de Artefactos: Generó y guardó exitosamente el informe de auditoría consolidado en el sistema de archivos: reports/wazuh_alerts_20261106-114422.pdf, incluyendo la tabla detallada con 10,000 alertas procesadas, IDs de regla (ej. 80792) y comandos auditados.

#### 2. Resultado Esperado

Se esperaba que el agente especialista en logs integrara la lectura activa del SIEM (Wazuh) con el marco conceptual del RAG (MITRE ATT&CK) para caracterizar un vector de ataque compuesto (acceso inicial por fuerza bruta seguido de una elevación de privilegios en Linux), finalizando con la exportación persistente de un reporte PDF listo para entrega ejecutiva o de auditoría.

#### 3. Por qué funcionó correctamente

* Orquestación Multi-paso (Grafo + Tools): El agente no intentó adivinar la información; alternó de forma autónoma entre consultar el motor vectorial FAISS y realizar llamadas de API a las herramientas del SOC (soc_tools).

* Precisión Semántica del RAG: El modelo de embeddings asoció la narrativa de la alerta (modificación de /etc/sudoers) con la técnica específica T1548.003, algo que una regla rígida de búsqueda léxica sin IA habría pasado por alto o catalogado de manera incompleta.

* Cierre de Ciclo Operativo: El agente completó el flujo integral desde la entrada en lenguaje natural hasta la persistencia física de un documento PDF con formato institucional en disco.
