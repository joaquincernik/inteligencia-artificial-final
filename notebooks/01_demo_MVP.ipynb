## 3. Consulta directa al RAG de MITRE ATT&CK (FAISS)

### Vertical: Threat Intelligence (threat_intel)

Esta prueba demuestra la capacidad de los agentes para recibir eventos o alertas directamente de Wazuh, buscar contexto enriquecido en la base de datos vectorial (FAISS) y ejecutar herramientas para profundizar en el análisis activo.

### Vertical: Threat Intelligence (threat_intel)

* `Prompt de validación:` "Dirígete al agente threat_intel. Analiza la alerta adjunta de Wazuh referida a la descarga y ejecución de un binario ELF sospechoso en el servidor Ubuntu. Utiliza el RAG para identificar tácticas de MITRE ATT&CK asociadas a botnets o rootkits de Linux. Posteriormente, ejecuta threat_tools para analizar el hash del binario y consultar la reputación de la IP de destino en bases de inteligencia. Genera un reporte de inteligencia en Markdown"  

* `Resultado esperado:`El agente debe enfocar su búsqueda en el RAG estrictamente en amenazas que afectan a distribuciones Linux, y luego usar sus herramientas para validar los indicadores de compromiso específicos (hashes de archivos ELF, IPs de C2), entregando un análisis contextualizado.

### Vertical: SOC Logs (soc_logs)

* `Prompt de validación:` "Dirígete al agente soc_logs. Examina las alertas de Wazuh del endpoint Ubuntu que indican múltiples fallos en auth.log seguidos de modificaciones en /etc/sudoers. Consulta en el RAG cómo mapear estos eventos a las técnicas de fuerza bruta y persistencia en Linux del framework MITRE ATT&CK. Luego, ejecuta soc_tools para extraer del SIEM los registros completos de auditd asociados a ese usuario durante las últimas 24 horas. Estructura la línea de tiempo del incidente en Markdown."   

* `Resultado esperado:` El agente debe interpretar los registros nativos del sistema operativo Ubuntu, recuperar el marco de referencia táctico del RAG y ejecutar las herramientas del SOC para reconstruir los pasos del atacante en el servidor
